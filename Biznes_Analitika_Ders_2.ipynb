## 6. Kənar dəyərlər: IQR metodu
**Kənar dəyər** (outlier) verilənlərin qalan hissəsi ilə müqayisədə qeyri-adi dərəcədə yüksək və ya aşağı olan dəyərdir.

- **IQR** (Kvartillərarası diapazon) = Q3 − Q1
- **Aşağı sərhəd** = Q1 − 1.5 × IQR
- **Yuxarı sərhəd** = Q3 + 1.5 × IQR

Sərhədlərdən kənarda qalan dəyərlər kənar dəyərlərdir.

```python
df_tip = df_1["tip"]
df_tip.describe()
```

```python
# Kvartillər, IQR və sərhədlər
Q1 = df_tip.quantile(0.25)
Q3 = df_tip.quantile(0.75)
IQR = Q3 - Q1

below = Q1 - 1.5 * IQR
above = Q3 + 1.5 * IQR
below, above
```

```python
# Bəxşişi kənar dəyər olan sətirlər
df_1[df_1["tip"] > above]
```

```python
# Yalnız sərhədlər daxilindəki dəyərləri saxlayır və sətirləri yenidən nömrələyirik
clear_df = df_1[df_1["tip"].between(below, above)].reset_index(drop=True)
clear_df['tip'].describe()
```
